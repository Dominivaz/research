```python
import os

def db2lin(db):
    return 10**(db / 10)

def load_sparams(path, prefix='ALL_', sparams=('S11', 'S12', 'S21', 'S22'), skiprows=18):
    first_file = os.path.join(path, f'{prefix}{sparams[0]}.csv')

    with open(first_file, 'r') as f:
        n_lines = sum(1 for _ in f)

    max_rows = n_lines - skiprows - 1

    freq = None
    values = []

    for sp in sparams:
        filename = os.path.join(path, f'{prefix}{sp}.csv')
        data = np.loadtxt(
            filename,
            delimiter=',',
            skiprows=skiprows,
            max_rows=max_rows
        )

        if freq is None:
            freq = data[:, 0]
        elif not np.allclose(data[:, 0], freq):
            raise ValueError(f"Frequency axis in {filename} does not match")

        values.append(data[:, 1])

    sparams_db = np.column_stack(values)
    sparams_lin = db2lin(sparams_db)

    return freq, sparams_db, sparams_lin

path = '/Users/dominiv/research/data/rfsoc_tests/components_sparams/'
freq, all_sparams_db, all_sparams_lin = load_sparams(path, prefix='ALL_')
freq, att_sparams_db, att_sparams_lin = load_sparams(path, prefix='ATT12_')
freq, amp_sparams_db, amp_sparams_lin = load_sparams(path, prefix='AMPS_')
freq, fil_sparams_db, fil_sparams_lin = load_sparams(path, prefix='FIL_')
```

```python
from eigsep_observing.io import read_hdf5
import numpy as np
import matplotlib.pyplot as plt

# Change this to your capture file
fname = "/Users/dominiv/eigsep/eigsep_observing/scripts/interweaved_comb.h5"

# Read file
data, header, metadata = read_hdf5(fname)

# Show basic contents
print("Pairs in file:", list(data.keys()))
print("\nData arrays:")
for k, v in data.items():
    print(f"  {k}: shape={v.shape}, dtype={v.dtype}")

print("\nHeader keys:", list(header.keys()))
print("Metadata keys:", list(metadata.keys()) if metadata else [])

# Pick a pair to plot
# Autos: "0", "1", "2", "3", "4", "5"
# Crosses: "02", "04", "13", "15", "24", "35"
pair = "0"

arr = data[pair]

# Use frequency axis if present; otherwise use channel number
freqs = header.get("freqs", np.arange(arr.shape[-1]))

# Convert frequency to MHz if it looks like Hz
if np.nanmax(freqs) > 1e6:
    x = freqs / 1e6
    xlabel = "Frequency [MHz]"
else:
    x = freqs
    xlabel = "Frequency"

# Plot first spectrum
plt.figure(figsize=(10, 4))
plt.plot(x, np.abs(arr[0]))
plt.xlabel(xlabel)
plt.ylabel(f"|{pair}|")
plt.title(f"{fname}: pair {pair}, spectrum 0")
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot average over all captured spectra
plt.figure(figsize=(10, 4))
plt.plot(x, np.abs(arr).mean(axis=0))
plt.xlabel(xlabel)
plt.ylabel(f"Mean |{pair}|")
plt.title(f"{fname}: pair {pair}, mean over {arr.shape[0]} spectra")
plt.grid(True)
plt.tight_layout()
plt.show()

# If this is a complex cross-pair, also plot phase
if np.iscomplexobj(arr):
    plt.figure(figsize=(10, 4))
    plt.plot(x, np.angle(arr[0]))
    plt.xlabel(xlabel)
    plt.ylabel("Phase [rad]")
    plt.title(f"{fname}: pair {pair}, phase spectrum 0")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python

# The three pairs you captured
pairs = ["0", "1", "13"]

# Read file
data, header, metadata = read_hdf5(fname)

print("Pairs in file:", list(data.keys()))

# Frequency axis
first_pair = pairs[0]
if first_pair not in data:
    first_pair = list(data.keys())[0]

arr0 = data[first_pair]
freqs = header.get("freqs", np.arange(arr0.shape[-1]))

# Convert Hz to MHz if needed
if np.nanmax(freqs) > 1e6:
    x = freqs / 1e6
    xlabel = "Frequency [MHz]"
else:
    x = freqs
    xlabel = "Frequency"

# Keep only pairs that are actually in the file
pairs_to_plot = [p for p in pairs if p in data]

if len(pairs_to_plot) == 0:
    raise ValueError(f"None of {pairs} were found in file. Found: {list(data.keys())}")

print("Plotting pairs:", pairs_to_plot)

# Plot first spectrum for each pair
plt.figure(figsize=(10, 5))

for pair in pairs_to_plot:
    arr = data[pair]
    plt.plot(x, np.abs(arr[0]), label=f"pair {pair}")

plt.xlabel(xlabel)
plt.ylabel("Amplitude")
plt.title(f"{fname}: first captured spectrum")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# Plot average over all captured spectra for each pair
plt.figure(figsize=(10, 5))

for pair in pairs_to_plot:
    arr = data[pair]
    plt.plot(x, np.abs(arr).mean(axis=0), label=f"pair {pair}")

plt.xlabel(xlabel)
plt.ylabel("Mean amplitude")
plt.title(f"{fname}: mean over captured spectra")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# Plot phase for complex pairs only, e.g. cross pair "13"
complex_pairs = [p for p in pairs_to_plot if np.iscomplexobj(data[p])]

if complex_pairs:
    plt.figure(figsize=(10, 5))

    for pair in complex_pairs:
        arr = data[pair]
        plt.plot(x, np.angle(arr[0]), label=f"pair {pair}")

    plt.xlabel(xlabel)
    plt.ylabel("Phase [rad]")
    plt.title(f"{fname}: phase of first spectrum")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
else:
    print("No complex cross-pairs found, so no phase plot was made.")
```
